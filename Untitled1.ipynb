```python
print('===== STUDENT PERFORMANCE ANALYZER =====\n')

# Collect student information
student_name = input('Enter student name: ')
student_age = int(input('Enter student age: '))
student_department = input('Enter student department: ')

# Collect scores
score_math = int(input('Enter score in Mathematics: '))
score_python = int(input('Enter score in Python: '))
score_statistics = int(input('Enter score in Statistics: '))

# Calculate total score
total_score = score_math + score_python + score_statistics

# Calculate average score (using type casting to float for accurate division)
average_score = float(total_score) / 3

# Determine highest and lowest scores
highest_score = max(score_math, score_python, score_statistics)
lowest_score = min(score_math, score_python, score_statistics)

# Determine pass/fail status (assuming passing score is >= 50 for all subjects)
# For simplicity, we'll assume a pass if the average is >= 50
passed = average_score >= 50

# Display results
print(f'\nName: {student_name}')
print(f'Age: {student_age}')
print(f'Department: {student_department}')

print(f'\nMathematics: {score_math}')
print(f'Python: {score_python}')
print(f'Statistics: {score_statistics}')
print(f'Total Score: {total_score}')
print(f'Average Score: {average_score:.1f}') # Format to one decimal place
print(f'Highest Score: {highest_score}')
print(f'Lowest Score: {lowest_score}')
print(f'Passed: {passed}')
```
