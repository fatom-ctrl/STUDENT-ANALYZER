```python
print("===== STUDENT PERFORMANCE ANALYZER =====")

name: str = "Alex"
age: int = 20
department: str = "Computer Science"

mathematics: float = 85
python_score: float = 92
statistics: float = 50

total_score: float = mathematics + python_score + statistics
average_score: float = total_score / 3

highest_score: float = max(mathematics, python_score, statistics)
lowest_score: float = min(mathematics, python_score, statistics)

passed: bool = average_score >= 50

print("\n===== STUDENT PERFORMANCE REPORT =====")
print(f"\nName: {name}")
print(f"Age: {age}")
print(f"Department: {department}")
print(f"\nMathematics: {mathematics:g}")
print(f"Python: {python_score:g}")
print(f"Statistics: {statistics:g}")
print(f"\nTotal Score: {total_score:g}")
print(f"Average Score: {average_score:g}")
print(f"Highest Score: {highest_score:g}")
print(f"Lowest Score: {lowest_score:g}")
print(f"Passed: {passed}")

num = "xyz"
print(num.upper())
```
