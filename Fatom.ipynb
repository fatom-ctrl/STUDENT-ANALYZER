```python
print("===== STUDENT PERFORMANCE ANALYZER =====")

name = "Alex"
age = 20
department = "Computer Science"

mathematics = 85
English = 66
python_score = 92
statistics = 50

total_score = mathematics + python_score + statistics + English
average_score = total_score / 3

# Alternative way to find highest and lowest scores without max/min
scores = [mathematics, python_score, statistics]
scores.sort()
highest_score = scores[-1]
lowest_score = scores[0]

passed = average_score >= 50

print("\n===== STUDENT PERFORMANCE REPORT =====")
print(f"\nName: {name}")
print(f"Age: {age}")
print(f"Department: {department}")
print(f"\nMathematics: {mathematics:g}")
print(f"English: {English:g}")
print(f"Python: {python_score:g}")
print(f"Statistics: {statistics:g}")
print(f"\nTotal Score: {total_score:g}")
print(f"Average Score: {average_score:g}")
print(f"Highest Score: {highest_score:g}")
print(f"Lowest Score: {lowest_score:g}")
print(f"Passed: {passed}")
```
